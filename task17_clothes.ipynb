```python
import pandas as pd
import matplotlib.pyplot as plt
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
data = pd.read_csv("Womens Clothing E-Commerce Reviews.csv")
```

Create Sentiment Score

```python
analyzer = SentimentIntensityAnalyzer()
data['Reviw Text'] = data['Review Text'].fillna("")
data["Sentiment_Score"] = data["Reviw Text"].apply(
        lambda x: analyzer.polarity_scores(x)["compound"]
)
```

```python
data.head()
```

Create Sentiment and Normalized Score

```python
data["Sentiment"] = data["Sentiment_Score"].apply(
    lambda x: "Positive" if x >= 0.05
    else "Negative" if x <= -0.05
    else "Neutral"
)

data["Normalized_Sentiment"] = (
    data["Sentiment_Score"] + 1
) / 2

data[["Review Text", "Sentiment_Score",
      "Sentiment", "Normalized_Sentiment"]].head()
```

```python
data.to_csv("clothing_sentiment.csv", index=False)
```
